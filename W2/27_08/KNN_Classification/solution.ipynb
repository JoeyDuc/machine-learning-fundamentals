```python
import pandas as pd
import numpy as np
from math import *
df = pd.read_csv('iris_1D.csv')
df
```

```python
# 1.Data
X = df['Petal_Length'].to_numpy()
print(X)
y = df['Label'].to_numpy()
print(y)
```

```python
x_test = 2.4
print(x_test)
k = 3
```

```python
distances = np.sqrt(((X - x_test) ** 2))
print(distances)
```

```python
distances_sorted = np.argsort(distances)
print(distances_sorted)
```

```python
k_nearest_neighbors = distances[distances_sorted[:k]]
print(k_nearest_neighbors)
```

```python
values = y[distances_sorted[:k]]
print(values)
```

```python
num_zeros = values.tolist().count(0)
num_ones = values.tolist().count(1)
print(num_zeros)
print(num_ones)
```

```python
df = pd.read_csv('iris_2D.csv')
df
```

```python
X = df[['Petal_Length', 'Petal_Width']].to_numpy()
X
```

```python
y = df['Label']
y
```

```python
x_test = [2.4, 0.8]
x_test = np.array(x_test)
```

```python
distances = np.sqrt(((X - x_test) ** 2).sum(axis = 1))
print(distances)
```

```python
distances_sorted = distances.argsort()
print(distances_sorted)
```

```python
k_nearest_neighbors = distances[distances_sorted[:k]]
print(k_nearest_neighbors)
```

```python
values = y[distances_sorted[:k]]
print(values)
```

```python
num_zeros = values.tolist().count(0)
num_ones = values.tolist().count(1)
print(num_zeros)
print(num_ones)
```

```python
data = pd.read_csv('iris_2D.csv')
x_data = data[['Petal_Length', 'Petal_Width']].to_numpy()
x_data = x_data.reshape(6, 2) 
y_data = data['Label'].to_numpy()

x_test = np.array([2.4, 0.8])
x_test = x_test.reshape(1, 2)
```

```python
from sklearn.neighbors import KNeighborsClassifier
classifier = KNeighborsClassifier(n_neighbors=1)
classifier.fit(x_data, y_data)
```

```python
y_pred = classifier.predict(x_test)
print(y_pred)
```

```python
classifier = KNeighborsClassifier(n_neighbors=6)
classifier.fit(x_data, y_data)
y_pred = classifier.kneighbors(x_test)
print(y_pred)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(x_data)
x_data = scaler.transform(x_data) #(x - mean) / (standard deviation)
print(x_data)
```
