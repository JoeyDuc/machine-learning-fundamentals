```python
k = 2
centroids = X[:k]
print(X)
print(centroids)
```

```python
inter = 10
for i in range(inter):
    # 2. Compute distances
    distances = np.abs(X.reshape(-1, 1) - centroids)
    print("distances:\n ", distances)
    
    # 3. Compute labels
    labels = np.argmin(distances, axis = 1)
    print("labels:\n ",labels)
    # 4. Update centroids
    centroids_new = np.array([X[labels == j].mean() for j in range(k)])
    #check
    if np.all(centroids == centroids_new):
        break
    centroids = centroids_new
    print("centroids:\n ", centroids)
wcss = np.sum([np.sum(np.square(X[labels == i] - centroids[i])) for i in range(2)])
print(wcss)
```

```python
age = np.array([18, 20, 22,
                30, 34, 40,
                60, 66, 70])
expenditure = np.array([80, 90, 85, 50, 64, 60, 30, 40, 25])
```

```python
X = np.column_stack((age, expenditure))
print(X)
```

```python
k = 3
max_inters = 2
centroids = X[2:k + 2]

for _ in range(max_inters):
    distances = np.linalg.norm(X[:, np.newaxis, :] - centroids, axis = 2)
    labels = np.argmin(distances, axis = 1)

    new_centroids = np.array([X[labels == i].mean(axis = 0) for i in range(k)])
    if np.all(centroids == new_centroids):
        break
    centroids = new_centroids
```
