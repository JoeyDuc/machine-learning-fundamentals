```python
import pandas as pd
df = pd.read_csv('ETTh1.csv')
df
```

```python
df['date'] = pd.to_datetime(df['date'])
df
```

```python
df.describe()
```

```python
import matplotlib.pyplot as plt
plt.plot(df.date[0:150], df.OT[0:150], marker = 'o', color = 'black', linewidth = 0.9, linestyle = '--',
         markeredgecolor = 'blue', markeredgewidth = 2.0, markerfacecolor = 'red', markersize =7.0)
plt.title('Oil Temperature', color = 'blue', size = 14)
plt.xlabel('Date', size = 14)
plt.ylabel('Value', size = 14)
plt.style.use('fivethirtyeight')
plt.grid(True)
plt.xticks(rotation = 90)
plt.show()
```

```python
import seaborn as sns
sns.lineplot(data = df, x = df.date[0:150], y = df.OT[0:150], marker='o', markersize = 10, markerfacecolor = 'green')
plt.xticks(rotation = 45)
```

```python
import plotly.express as px
fig = px.line(df[0:150], x = 'date', y = 'OT', title = 'Oil Temperature', markers = True)
fig.update_layout(
    xaxis_title = "Date", yaxis_title = 'Value'
)
fig.show()
```

```python
fig, axes = plt.subplots(nrows = 7, ncols = 1, figsize = (15, 20))

for i, column in enumerate(df.columns[1:]):
    axes[i].plot(df['date'], df[column], label = column)
    axes[i].set_title(column)
    axes[i].set_xlabel('Date')
    axes[i].set_ylabel('Values')
    axes[i].legend()
plt.tight_layout()
plt.show()
```

```python
#selecting a time period to zoom in on
start = 0
end = 150

#Creating the main line chart with a highlighted square
fig, ax = plt.subplots(figsize = (12, 7))

#Plotting the main line chart for a selected variable (example: 'HUFL')
ax.plot(df['date'], df['HUFL'], label = 'HUFL')

#Adding a rectangble to highlight the zoomed-in area
rect = plt.Rectangle((df['date'][start], min(df['HUFL'])),
                     df['date'][end] - df['date'][start],
                     max(df['HUFL']) - min(df['HUFL']),
                     linewidth = 1, edgecolor = 'r', facecolor = 'none'
                     )

ax.add_patch(rect)

#Creating an inset (zoomed-in) plot within the main plot
ax_inset = fig.add_axes([0.5, 0.5, 0.35, 0.35]) #[lef, bottom, width, height]
ax_inset.plot(df['date'][start:end], df['HUFL'][start:end], color = 'r')
ax_inset.set_title('Zoomed-in Period')

#Setting labels and title for the main plot

ax.set_xlabel('Date')
ax.set_ylabel('HUFL')
ax.set_title('Line Chart with Zoomed-in Period')

#Display
plt.show()
```

```python
peak_values = df.max()
peak_times = df.idxmax()

peak_analysis = pd.DataFrame({'Peak Value':peak_values, 'Peak Time': peak_times})
print(peak_analysis)
peak_analysis = peak_analysis.drop('date')
print(peak_analysis)

plt.figure(figsize=(12, 6))
peak_analysis['Peak Value'].plot(kind = 'bar', color = 'skyblue')
plt.title('Peak Values for Each Load Type and Oil Temperature')
plt.xlabel('Variable')
plt.ylabel('Peak Value')
plt.xticks(rotation = 45)
plt.show()
```

```python
import numpy as np
counts = {column : [
    np.sum(df[column] < 0),
    np.sum(df[column] == 0),
    np.sum(df[column] > 0)]
    for column in df.columns[1:]
}

counts_df = pd.DataFrame(counts, index = ['Negative', 'Zeros', 'Positives']).T

ax = counts_df.plot(kind = 'bar', stacked = True, figsize = (12, 8))
ax.set_title('count of Zeros, Negative, and Positive Values for Each Variable')
ax.set_xlabel('Variables')
ax.set_ylabel('Counts')
plt.xticks(rotation = 45)
plt.legend(title = 'Value Type')
plt.tight_layout()
plt.show()
```

```python
#Donut chart
#Preparing the data for the stacked bar chart to count zeros, negatives, positives for each variable
counts = {column : [
    np.sum(df[column] < 0),
    np.sum(df[column] == 0),
    np.sum(df[column] > 0)
] for column in df.columns[1:]}

#Creating a DataFrame from the counts dictionary

counts_for_donut = counts_df.sum()
 #Plotting the donut chart
fig, ax = plt.subplots()
ax.pie(counts_for_donut, labels = counts_for_donut.index, autopct = '%.1f%%', startangle=90, wedgeprops=dict(width = 0.3))

#Draaw a circle at the center to create a donut hole
centre_circle = plt.Circle((0, 0), 0.70, fc = 'white')
fig.gca().add_artist(centre_circle)

#Equal aspect ratic ensures that pie is drawn as a circle.

ax.axis('equal')

plt.title('Count of Zeros, Negatives, and Positives Values Across All Variables')
plt.show()
```
