```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
df = pd.read_csv('StudentsPerformance.csv')
df
```

```python
#Student Performance Data from Kaggle
fig = plt.figure(figsize = (4, 5))
plt.style.use('fivethirtyeight')
plt.bar(
    x = df['race/ethnicity'].unique(),
    height = df['race/ethnicity'].value_counts().to_list(),
    color = 'blue', #Set the color of the bars
    edgecolor = 'black', #Set the color of the bar edges
    linewidth = 1.5, # Set the width of the bar edges
    alpha = 0.8 #Set the transparency of the bars
)
plt.xlabel('Race/Enthncity', fontsize = 14, fontweight = 'bold') # Set the x-axis label with font size and style

plt.ylabel('Count', fontsize = 14, fontweight = 'bold') # Set the y-axis label font size and style
plt.title('Race/Ethncity Counts', fontsize = 16, fontweight = 'bold') # Set the chart title with font size and style
plt.xticks(fontsize = 12, rotation = 45) #Set the font size of the x-axis tick labels
plt.yticks(fontsize = 12) # Set the font size of the y-axis tick labels
plt.grid(True, linestyle='--', linewidth = 0.5, alpha = 0.5, color = 'black') # Display gridlines with a dashed style and reduced opacity
plt.show()
```

```python
# Đếm số lượng male/female theo từng nhóm race/ethnicity
grouped_data = (
    df.groupby(["race/ethnicity", "gender"])
        .size()
        .unstack(fill_value=0)
)

print(grouped_data)

# Vẽ biểu đồ
fig = plt.figure(figsize=(4, 5))

plt.style.use("fivethirtyeight")

# Female
plt.bar(
    x=grouped_data.index,
    height=grouped_data["female"],
    label="Female",
    color="red",
    edgecolor="black",
    linewidth=1.5,
    alpha=0.8
)

# Male - chồng lên Female
plt.bar(
    x=grouped_data.index,
    height=grouped_data["male"],
    bottom=grouped_data["female"],
    label="Male",
    color="green",
    edgecolor="black",
    linewidth=1.5,
    alpha=0.8
)

plt.xlabel(
    "Race/Ethnicity",
    fontsize=14,
    fontweight="bold"
)

plt.ylabel(
    "Count",
    fontsize=14,
    fontweight="bold"
)

plt.title(
    "Race/Ethnicity Counts",
    fontsize=16,
    fontweight="bold"
)

plt.xticks(
    fontsize=12,
    rotation=45
)

plt.yticks(fontsize=12)

plt.grid(
    True,
    linestyle="--",
    linewidth=0.5,
    alpha=0.5,
    color="black"
)

plt.legend()

plt.show()
```
