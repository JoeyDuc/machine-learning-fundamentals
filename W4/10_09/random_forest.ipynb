```python
import pandas as pd
import numpy as np

from sklearn.ensemble import RandomForestClassifier
```

```python
# load data
df = pd.read_csv('data/iris_2D.csv')
df
```

```python
# Get data
x_data = df[['Petal_Length', 'Petal_Width']].to_numpy()
x_data = x_data.reshape(6, 2)
x_data
```

```python
y_data = df['Label'].to_numpy(dtype = np.uint8)
y_data
```

```python
# Define model
rf_classifier = RandomForestClassifier(
    n_estimators=3,
    max_features=1,
    max_depth=1,
    criterion='entropy', max_samples=6)

#Train model
rf_classifier.fit(x_data, y_data)
```

```python
x_test = np.array([[2.7, 0.8]])
y_predicted = rf_classifier.predict(x_test)
y_predicted
```

```python
# Visualization

from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
plt.figure(figsize=(6, 6))
_ = plot_tree(rf_classifier.estimators_[0],
              feature_names=['Petal_Length', 'Petal_Width'],
              filled=True)
```

```python
plt.figure(figsize=(6, 6))
_ = plot_tree(rf_classifier.estimators_[1],
              feature_names=['Petal_Length', 'Petal_Width'],
              filled=True)
```

```python
plt.figure(figsize=(6, 6))
_ = plot_tree(rf_classifier.estimators_[2],
              feature_names=['Petal_Length', 'Petal_Width'],
              filled=True)
```
