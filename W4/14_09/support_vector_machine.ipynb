```python
import numpy as np
from sklearn.svm import SVC

X = np.array([
    [3, 4], [1, 4], [2, 3], 
    [6, -1], [7, -1], [5, -3]
])
y = np.array([
    -1, -1, -1, 1, 1, 1
])

clf = SVC(kernel = 'linear')
clf.fit(X, y)

w = clf.coef_
b = clf.intercept_
print(w)

print(b)
```

```python
X_test = np.array([
    [8, -2],
    [1, 3]
])
y_pred = clf.predict(X_test)
print(y_pred)
```
