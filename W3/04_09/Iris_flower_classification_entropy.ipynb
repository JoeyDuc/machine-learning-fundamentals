```python
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn import tree


# 1. Load dataset Iris
dataset = load_iris()

X = dataset.data
y = dataset.target


# 2. Tạo mô hình Decision Tree
classifier = tree.DecisionTreeClassifier(
    criterion="entropy",
    max_depth=4,
    min_samples_leaf=10,
    random_state=42
)


# 3. Huấn luyện mô hình
classifier.fit(X, y)


# 4. Vẽ cây quyết định
fig, ax = plt.subplots(figsize=(10, 10))

tree.plot_tree(
    classifier,
    ax=ax,
    feature_names=[
        "sepal length",
        "sepal width",
        "petal length",
        "petal width"
    ],
    class_names=dataset.target_names,
    filled=True
)

plt.show()
```

```python
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn import tree


# 1. Load dataset Iris
dataset = load_iris()

X = dataset.data
y = dataset.target


# 2. Tạo mô hình Decision Tree
classifier = tree.DecisionTreeClassifier(
    criterion="gini",
    max_depth=4,
    min_samples_leaf=10,
    random_state=42
)


# 3. Huấn luyện mô hình
classifier.fit(X, y)


# 4. Vẽ cây quyết định
fig, ax = plt.subplots(figsize=(10, 10))

tree.plot_tree(
    classifier,
    ax=ax,
    feature_names=[
        "sepal length",
        "sepal width",
        "petal length",
        "petal width"
    ],
    class_names=dataset.target_names,
    filled=True
)

plt.show()
```
