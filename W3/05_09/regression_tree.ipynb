```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# High Bias: kết quả dự đoán và kết quả thực tế nó quá khác => bias (training error lớn)
# Low Bias: Kết quả dự đoán và kết quả thực tế thấp (training error thấp)
```

```python
# Low Variance: thể hiện training error và test error không khác nhau: train và test đều cao hoặc đều thấp
# High Variance: thể hiện training error và test error sai lệch lớn: train thấp - test cao, train cao - test thấp
```

```python
# Underfit: high bias và low variance
# Good Fit: 
# Overfit: low bias và high variance
```

```python
df = pd.read_csv('MSFT.csv')
df
```

```python
plt.figure(figsize = (10, 4))
plt.title('Microsoft Stock Prices')
plt.xlabel('Date')
plt.ylabel('Close')
plt.plot(df['Close'])
plt.show()
```

```python
x = df[['Open', 'High', 'Low']]
y = df['Close']
x = x.to_numpy()
y = y.to_numpy()
y = y.reshape(-1, 1)
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.2, random_state = 42)

from sklearn.tree import DecisionTreeRegressor
model = DecisionTreeRegressor()
model.fit(xtrain, ytrain)
ypred = model.predict(xtest)
df = pd.DataFrame(data = {'Predicted Rate' : ypred})
print(df.head())
```
