```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv('iris.csv')
data
```

```python
data['species'].value_counts()
```

```python
pd.plotting.parallel_coordinates(data, "species")
```

```python
sns.jointplot(x = 'sepal_length', y = 'sepal_width', data = data, hue = 'species')
plt.show()
```

```python
sns.jointplot(x = 'petal_length', y = 'petal_width', data = data, hue = 'species')
plt.show()
```

```python
sns.stripplot(y = 'sepal_length', x = 'species', data = data, hue = 'species')
plt.show()
```

```python
sns.stripplot(y = 'sepal_width', x = 'species', data = data, hue = 'species')
plt.show()
```

```python
sns.boxplot(x = 'species', y = 'sepal_width', data = data, hue = 'species')
plt.show()
```
